```python
import pandas as pd
import os
defense=[]
for year in range(2017,2026):
    frame = pd.read_csv(f'scraped_data/{year}_dfgtotal.csv')
    defense.append(frame)
dfga= pd.concat(defense)
dfga
def load_shot_data_for_teams(main_df):
    """
    Iterates through each team_id/year combination from the main DataFrame,
    reads the corresponding regular season and post-season shot data CSVs,
    and stores the combined data in a dictionary.

    Args:
        main_df (pd.DataFrame): The main DataFrame containing 'team_id' and 'year' columns.

    Returns:
        dict: A dictionary where keys are (team_id, year) tuples and values
              are DataFrames containing combined regular and post-season data.
    """
    # Get unique combinations of team_id and year
    unique_teams_years = main_df[['team_id', 'year']].drop_duplicates()
    
    all_shot_data = []
    base_path = '../../shot_data/team'
    
    print(f"Found {len(unique_teams_years)} unique team/year combinations to process.")
    
    # Iterate over each unique row (combination)
    for index, row in unique_teams_years.iterrows():
        team_id = row['team_id']
        year = row['year']
        
        season_dfs = []

        # --- Load Regular Season Data ---
        reg_season_path = os.path.join(base_path, str(year), f'{team_id}.csv')
        try:
            reg_df = pd.read_csv(reg_season_path)
            reg_df['season_type'] = 'REG'
            season_dfs.append(reg_df)
            print(f"Successfully loaded REGULAR season data for Team ID: {team_id}, Year: {year}")
        except FileNotFoundError:
            print(f"Info: Regular season file not found for Team ID: {team_id}, Year: {year} at path: {reg_season_path}")
        except Exception as e:
            print(f"An error occurred reading regular season file for Team ID: {team_id}, Year: {year}. Error: {e}")

        # --- Load Postseason Data ---
        post_season_path = os.path.join(base_path, f'{year}ps', f'{team_id}.csv')
        try:
            post_df = pd.read_csv(post_season_path)
            post_df['season_type'] = 'PS'
            season_dfs.append(post_df)
            print(f"Successfully loaded POST season data for Team ID: {team_id}, Year: {year}")
        except FileNotFoundError:
            print(f"Info: Postseason file not found for Team ID: {team_id}, Year: {year} at path: {post_season_path}")
        except Exception as e:
            print(f"An error occurred reading postseason file for Team ID: {team_id}, Year: {year}. Error: {e}")

        # --- Combine and Store ---
        if season_dfs:
            combined_df = pd.concat(season_dfs)
            combined_df['year']=year
            all_shot_data.append(combined_df)
        
        else:
            print(f"Warning: No data found at all for Team ID: {team_id}, Year: {year}")
            
    print("\nFinished processing all team/year combinations.")
    return pd.concat(all_shot_data)
shot_data=load_shot_data_for_teams(dfga)
shot_data
```

```python
import pandas as pd

# --- Assume 'dfga' and 'shot_data' are already loaded from your previous steps ---
# dfga: DataFrame with at least ['gi', 'ei', 'def_id'] (one def_id per row)
# shot_data: The main DataFrame with all shot details.

# 1. Load and prepare the LEBRON defensive stats data
lebron_url = 'https://raw.githubusercontent.com/gabriel1200/site_Data/refs/heads/main/lebron.csv'
lebron_df = pd.read_csv(lebron_url)

# Select and rename the necessary columns
lebron_df = lebron_df[['Pos', 'Defensive Role', 'NBA ID', 'D-LEBRON']]
lebron_df = lebron_df.rename(columns={
    'Pos': 'DEF_POSITION',
    'Defensive Role': 'DEF_ROLE',
    'NBA ID': 'def_id',  # Corrected column name for the merge
    'D-LEBRON': 'D_LEBRON'
})

# 2. Merge LEBRON stats onto the original dfga DataFrame
# Ensure the merge keys are of the same data type.
dfga['def_id'] = pd.to_numeric(dfga['def_id'], errors='coerce')
dfga.dropna(subset=['def_id'], inplace=True) # Drop rows where def_id is not a number
dfga['def_id'] = dfga['def_id'].astype(int)

lebron_df['def_id'] = lebron_df['def_id'].astype(int)

# Left merge to add defensive stats to each defender instance
dfga_with_stats = pd.merge(dfga, lebron_df, on='def_id', how='left')


# 3. Group by game/event ID and aggregate all relevant columns
# Convert columns to be aggregated into strings to handle NaNs and allow joining
agg_cols = ['def_id', 'DEF_POSITION', 'DEF_ROLE', 'D_LEBRON']
for col in agg_cols:
    dfga_with_stats[col] = dfga_with_stats[col].astype(str)

# Define the aggregation for each column
agg_functions = {
    'def_id': '|'.join,
    'DEF_POSITION': '|'.join,
    'DEF_ROLE': '|'.join,
    'D_LEBRON': '|'.join
}

# Group by 'gi' and 'ei' and apply the aggregation
df_combined = dfga_with_stats.groupby(['gi', 'ei']).agg(agg_functions).reset_index()


# 4. Finalize the merge with the main shot data
# Rename columns for clarity before the final merge
df_combined.rename(columns={
    'gi': 'GAME_ID',
    'ei': 'GAME_EVENT_ID',
    'def_id': 'DEF_ID' # Note: The other columns are already named correctly
}, inplace=True)

# Merge the combined defensive data into the main shot_data DataFrame
merged_data = shot_data.merge(df_combined, on=['GAME_ID', 'GAME_EVENT_ID'], how='left')

# Print the head to verify the new columns
print("Sample of merged data with new defensive columns:")
print(merged_data[['GAME_ID', 'GAME_EVENT_ID', 'DEF_ID', 'DEF_POSITION', 'DEF_ROLE', 'D_LEBRON']].head())

# 5. Print the percentage of shots with logged defenders (as before)
print("\n--- Coverage Stats ---")
for year in range(2018, 2026):
    year_data = merged_data[merged_data.year == year]
    if not year_data.empty:
        coverage = 100 - (100 * year_data['DEF_ID'].isna().sum() / len(year_data))
        print(f"For {year}, {coverage:.2f}% of shots have logged defenders")
```

```python
merged_data.columns
def save_merged_data_by_team_year(merged_df, base_path='team'):
    """
    Saves the merged shot data back into a structured directory,
    split by team, year, and season type, sorted by game events.

    Args:
        merged_df (pd.DataFrame): The DataFrame containing all merged shot data.
                                  Must include 'team_id', 'year', 'season_type',
                                  'GAME_ID', and 'GAME_EVENT_ID' columns.
        base_path (str): The base directory to save the files into.
    """
    if not os.path.exists(base_path):
        os.makedirs(base_path)
        print(f"Created base directory for saving: {base_path}")

    # Get unique combinations of team_id and year
    unique_teams_years = merged_df[['TEAM_ID', 'year']].drop_duplicates()
    
    print(f"\nFound {len(unique_teams_years)} unique team/year combinations to save.")

    for index, row in unique_teams_years.iterrows():
        team_id = row['TEAM_ID']
        year = row['year']

        # Filter data for the current team and year
        team_year_data = merged_df[(merged_df['TEAM_ID'] == team_id) & (merged_df['year'] == year)]

        # --- Process and Save Regular Season Data ---
        reg_data = team_year_data[team_year_data['season_type'] == 'REG'].copy()
        if not reg_data.empty:
            # Sort the data
            reg_data.sort_values(by=['GAME_ID', 'GAME_EVENT_ID'], inplace=True)
            
            # Define path and create directory
            reg_season_dir = os.path.join(base_path, str(year))
            os.makedirs(reg_season_dir, exist_ok=True)
            reg_season_path = os.path.join(reg_season_dir, f'{team_id}.csv')
            
            # Save to CSV
            reg_data.to_csv(reg_season_path, index=False)
            print(f"Saved REGULAR season data to: {reg_season_path}")

        # --- Process and Save Postseason Data ---
        ps_data = team_year_data[team_year_data['season_type'] == 'PS'].copy()
        if not ps_data.empty:
            # Sort the data
            ps_data.sort_values(by=['GAME_ID', 'GAME_EVENT_ID'], inplace=True)
            
            # Define path and create directory
            post_season_dir = os.path.join(base_path, f'{year}ps')
            os.makedirs(post_season_dir, exist_ok=True)
            post_season_path = os.path.join(post_season_dir, f'{team_id}.csv')
            
            # Save to CSV
            ps_data.to_csv(post_season_path, index=False)
            print(f"Saved POST season data to: {post_season_path}")

    print("\nFinished saving all merged data.")
save_merged_data_by_team_year(merged_data)
```

```python
def save_merged_data_by_defender(merged_df, base_path=''):
    """
    Saves the merged shot data into a directory structured by defender,
    year, and season type. Handles multi-defender events.

    Args:
        merged_df (pd.DataFrame): The DataFrame containing all merged shot data.
                                  Must include 'DEF_ID', 'year', 'season_type',
                                  'GAME_ID', and 'GAME_EVENT_ID' columns.
        base_path (str): The base directory to save the files into.
    """
    if not os.path.exists(base_path):
        os.makedirs(base_path)
        print(f"Created base directory for saving: {base_path}")

    # --- Pre-processing to get all unique, individual defender IDs ---
    # Drop rows where DEF_ID is NaN, as we can't process them.
    df_filtered = merged_df.dropna(subset=['DEF_ID']).copy()
    df_filtered['DEF_ID'] = df_filtered['DEF_ID'].astype(str)

    # Split the DEF_ID string on '|' and use explode to give each ID its own row
    s = df_filtered['DEF_ID'].str.split('|').explode()
    unique_defender_ids = s.unique()
    
    print(f"\nFound {len(unique_defender_ids)} unique defender IDs to process and save.")

    # --- Iterate through each unique defender and save their data ---
    for def_id in unique_defender_ids:
        # Filter the original dataframe to find all shots contested by this defender.
        # Using word boundaries (\b) ensures we match the whole ID (e.g., '123' not '1234').
        defender_data = df_filtered[df_filtered['DEF_ID'].str.contains(f'\\b{def_id}\\b', regex=True)]

        # Find all years this defender has data for
        years_for_defender = defender_data['year'].unique()

        for year in years_for_defender:
            # Filter for the specific year
            defender_year_data = defender_data[defender_data['year'] == year]

            # --- Process and Save Regular Season Data ---
            reg_data = defender_year_data[defender_year_data['season_type'] == 'REG'].copy()
            if not reg_data.empty:
                reg_data.sort_values(by=['GAME_ID', 'GAME_EVENT_ID'], inplace=True)
                reg_dir = os.path.join(base_path, str(year))
                os.makedirs(reg_dir, exist_ok=True)
                reg_path = os.path.join(reg_dir, f'{def_id}.csv')
                reg_data.to_csv(reg_path, index=False)
                print(f"Saved REG data for Defender {def_id}, Year {year} to: {reg_path}")

            # --- Process and Save Postseason Data ---
            ps_data = defender_year_data[defender_year_data['season_type'] == 'PS'].copy()
            if not ps_data.empty:
                ps_data.sort_values(by=['GAME_ID', 'GAME_EVENT_ID'], inplace=True)
                ps_dir = os.path.join(base_path, f'{year}ps')
                os.makedirs(ps_dir, exist_ok=True)
                ps_path = os.path.join(ps_dir, f'{def_id}.csv')
                ps_data.to_csv(ps_path, index=False)
                print(f"Saved PS data for Defender {def_id}, Year {year} to: {ps_path}")

    print("\nFinished saving all data by defender.")
save_merged_data_by_defender(merged_data)
```
