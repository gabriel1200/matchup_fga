```python
import pandas as pd
import os
defense=[]
for year in range(2017,2026):
    frame = pd.read_csv(f'scraped_data/{year}_dfgtotal.csv')
    defense.append(frame)
dfga= pd.concat(defense)
dfga
def load_shot_data_for_teams(main_df):
    """
    Iterates through each team_id/year combination from the main DataFrame,
    reads the corresponding regular season and post-season shot data CSVs,
    and stores the combined data in a dictionary.

    Args:
        main_df (pd.DataFrame): The main DataFrame containing 'team_id' and 'year' columns.

    Returns:
        dict: A dictionary where keys are (team_id, year) tuples and values
              are DataFrames containing combined regular and post-season data.
    """
    # Get unique combinations of team_id and year
    unique_teams_years = main_df[['team_id', 'year']].drop_duplicates()
    
    all_shot_data = []
    base_path = '../../shot_data/team'
    
    print(f"Found {len(unique_teams_years)} unique team/year combinations to process.")
    
    # Iterate over each unique row (combination)
    for index, row in unique_teams_years.iterrows():
        team_id = row['team_id']
        year = row['year']
        
        season_dfs = []

        # --- Load Regular Season Data ---
        reg_season_path = os.path.join(base_path, str(year), f'{team_id}.csv')
        try:
            reg_df = pd.read_csv(reg_season_path)
            reg_df['season_type'] = 'REG'
            season_dfs.append(reg_df)
            print(f"Successfully loaded REGULAR season data for Team ID: {team_id}, Year: {year}")
        except FileNotFoundError:
            print(f"Info: Regular season file not found for Team ID: {team_id}, Year: {year} at path: {reg_season_path}")
        except Exception as e:
            print(f"An error occurred reading regular season file for Team ID: {team_id}, Year: {year}. Error: {e}")

        # --- Load Postseason Data ---
        post_season_path = os.path.join(base_path, f'{year}ps', f'{team_id}.csv')
        try:
            post_df = pd.read_csv(post_season_path)
            post_df['season_type'] = 'PS'
            season_dfs.append(post_df)
            print(f"Successfully loaded POST season data for Team ID: {team_id}, Year: {year}")
        except FileNotFoundError:
            print(f"Info: Postseason file not found for Team ID: {team_id}, Year: {year} at path: {post_season_path}")
        except Exception as e:
            print(f"An error occurred reading postseason file for Team ID: {team_id}, Year: {year}. Error: {e}")

        # --- Combine and Store ---
        if season_dfs:
            combined_df = pd.concat(season_dfs)
            combined_df['year']=year
            all_shot_data.append(combined_df)
        
        else:
            print(f"Warning: No data found at all for Team ID: {team_id}, Year: {year}")
            
    print("\nFinished processing all team/year combinations.")
    return pd.concat(all_shot_data)
shot_data=load_shot_data_for_teams(dfga)
shot_data
```

```python
# First, ensure the 'def_id' column is treated as a string
dfga['def_id'] = dfga['def_id'].astype(str)

# Group by 'gi' and 'ei', then aggregate 'def_id' by joining the strings
df_combined = dfga.groupby(['gi', 'ei'])['def_id'].apply('|'.join).reset_index()

# Now, df_combined has unique pairs of gi/ei, with the def_id's combined.
print(df_combined.head())
dfga=df_combined.copy()

dfga.rename(columns={'gi':'GAME_ID','ei':'GAME_EVENT_ID','def_id':'DEF_ID'},inplace=True)
merged_data=shot_data.merge(dfga,how='left')
for year in range(2018,2026):
    year_data=merged_data[merged_data.year==year]
    print(f"For {year}, {100 - (100 * year_data['DEF_ID'].isna().sum()/len(year_data))}% of shots have logged defenders ")
```

```python
merged_data.columns
def save_merged_data_by_team_year(merged_df, base_path='team'):
    """
    Saves the merged shot data back into a structured directory,
    split by team, year, and season type, sorted by game events.

    Args:
        merged_df (pd.DataFrame): The DataFrame containing all merged shot data.
                                  Must include 'team_id', 'year', 'season_type',
                                  'GAME_ID', and 'GAME_EVENT_ID' columns.
        base_path (str): The base directory to save the files into.
    """
    if not os.path.exists(base_path):
        os.makedirs(base_path)
        print(f"Created base directory for saving: {base_path}")

    # Get unique combinations of team_id and year
    unique_teams_years = merged_df[['team_id', 'year']].drop_duplicates()
    
    print(f"\nFound {len(unique_teams_years)} unique team/year combinations to save.")

    for index, row in unique_teams_years.iterrows():
        team_id = row['team_id']
        year = row['year']

        # Filter data for the current team and year
        team_year_data = merged_df[(merged_df['team_id'] == team_id) & (merged_df['year'] == year)]

        # --- Process and Save Regular Season Data ---
        reg_data = team_year_data[team_year_data['season_type'] == 'REG'].copy()
        if not reg_data.empty:
            # Sort the data
            reg_data.sort_values(by=['GAME_ID', 'GAME_EVENT_ID'], inplace=True)
            
            # Define path and create directory
            reg_season_dir = os.path.join(base_path, str(year))
            os.makedirs(reg_season_dir, exist_ok=True)
            reg_season_path = os.path.join(reg_season_dir, f'{team_id}.csv')
            
            # Save to CSV
            reg_data.to_csv(reg_season_path, index=False)
            print(f"Saved REGULAR season data to: {reg_season_path}")

        # --- Process and Save Postseason Data ---
        ps_data = team_year_data[team_year_data['season_type'] == 'PS'].copy()
        if not ps_data.empty:
            # Sort the data
            ps_data.sort_values(by=['GAME_ID', 'GAME_EVENT_ID'], inplace=True)
            
            # Define path and create directory
            post_season_dir = os.path.join(base_path, f'{year}ps')
            os.makedirs(post_season_dir, exist_ok=True)
            post_season_path = os.path.join(post_season_dir, f'{team_id}.csv')
            
            # Save to CSV
            ps_data.to_csv(post_season_path, index=False)
            print(f"Saved POST season data to: {post_season_path}")

    print("\nFinished saving all merged data.")
save_merged_data_by_team_year(merged_data)
```
